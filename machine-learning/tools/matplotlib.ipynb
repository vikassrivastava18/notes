```python
import matplotlib.pyplot as plt
```

```python
x = [1,2,3,4,5]
y = [85,67,95,56,73]
```

### Creating a line graph

```python
plt.plot(x, y)
plt.title("Performance Report")
plt.xlabel("Semester")
plt.ylabel("Percentage of marks")
```

### Creating a bar graph

```python
plt.bar(x, y)
plt.title("Performance Report")
plt.xlabel("Semester")
plt.ylabel("Percentage of marks")
```

### Histogram

```python
x = [1,1,3,4,5,6,7,7]
bins = [1,2,3,4,5,6,7]
plt.hist(x, bins=bins)
plt.title("Histogram")
plt.legend(["bar"])
```

### Pie Chart

```python
x = [1, 2, 3, 4]
e = (0, 0 , 0.1, 0)

plt.pie(x, explode=e)

plt.title("Pie Chart")
```

### 3-D graph

```python
x = [1,2,3,4,5]
y = [1,4,9,16,25]
z = [1,8,27,64,125]

fig = plt.figure()
ax = plt.axes(projection='3d')
ax.plot3D(z,y,x)
```
