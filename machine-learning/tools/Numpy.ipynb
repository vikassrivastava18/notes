# Vector Operations

```python
a = np.arange(10)
print(a)
print(type(a))

# access an element
print(f"a[2].shape: {a[2].shape} a[2] = {a[2]}")
print(a[-1])

try:
    c = a[10]
except Exception as e:
    print("Error: ", e)

c = a[2:7:1];   print(f"a[2:7:2] = {c}")
d = a[2:7:2]; print(f"a[2:7:2] = {d}") # Steps
```

```python
a = np.array([1,2,3,4])
print(f"a             : {a}")
# negate elements of a
b = -a
print(f"b = -a        : {b}")

# sum all elements of a, returns a scalar
b = np.sum(a)
print(f"b = np.sum(a) : {b}")

b = np.mean(a)
print(f"b = np.mean(a): {b}")

b = a**2
print(f"b = a**2      : {b}")

b = np.max(a)
print(f"b = np.max(a)  : {b}")

b = np.square(a)
print(f"b = np.square(a) = {b}")
```

```python
a = np.array([ 1, 2, 3, 4])
b = np.array([-1,-2, 3, 4])
print(f"Binary operators work element wise: {a + b}")
print(f"np.add(a, b) = {np.add(a, b)}")
```

# Scalar Vector Operations

```python
a = np.array([1,2,3,4])
print(f"shape of a: {a.shape}, type: {type(a)}")
b = 5 * a
print(b)
c = a * 5
print(c)
```

```python
# Dot Product
def my_dot(a, b):
    dot = 0
    for i in range(a.shape[0]):
        dot += a[i] * b [i]
    return dot
a = np.array([1,2,3,4])
b = np.array([1,1,1,1])
print(f"Dot product of a and b: {my_dot(a, b)}")
```

```python
a = np.array([1,2,3,4])
b = np.array([1,1,1,1])
d = np.dot(a, b)
print(f"Dot product of a and b: {d}")
```

# Vectorization

```python
import time

np.random.seed(1)
a = np.random.rand(100000000) # Very large arrays
b = np.random.rand(100000000)
print(a, b)
tic = time.time()
c = np.dot(a, b)
toc = time.time()
print(f"Vectorized version duration: {1000*(toc-tic):.4f} ms")

tic = time.time()
c = my_dot(a, b)
toc = time.time()
print(f"loop version duration: {1000*(toc-tic):.4f} ms")
```

# Broadcasting example

```python
A = np.array([[56.0, 0.0, 4.4, 68.0],
              [1.2, 104.0, 52.0, 8.0],
              [1.8, 135.0, 99.0, 0.9]])
print(A)
```

```python
cal = A.sum(axis=0) # Sum the columns vertically, = 1 for horizontal summation
print(cal)
print(cal.reshape(1,4))

_mean = A.mean(axis=0)
print(f"Mean of columns")
print(f"{_mean}")
```

```python
total_calories = cal.reshape(1, 4)
percentage = 100 * A / total_calories
print(percentage)
```

# Normalization

```python
x = np.array([[0., 3., 4.],
              [1., 6., 4.]])
# axis=1 => horizontal, axis=0 => vertical
x_norm = np.linalg.norm(x, axis=1)
print(x_norm)
x_norm = np.linalg.norm(x, axis=1, keepdims=True)
print(x_norm)
print(x / x_norm)
```

# Softmax



```python
t_x = np.array([[9, 2, 5, 0, 0],
                [7, 5, 0, 0 ,0]])

x_exp = np.exp(t_x)
x_sum = np.sum(x_exp, axis=1, keepdims=True)
x_softmax = x_exp / x_sum
print(x_softmax)
```
